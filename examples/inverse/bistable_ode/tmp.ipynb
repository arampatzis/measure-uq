```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp

def ode_rhs(t, y, r):
    return r * (y - 1) * (2 - y) * (y - 3)

r = 1
y0_values = [0, 1, 1.5, 2, 2.5, 3, 4]
t_span = (0, 8)
t_eval = np.linspace(t_span[0], t_span[1], 300)

plt.figure(figsize=(8, 6))
for y0 in y0_values:
    sol = solve_ivp(ode_rhs, t_span, [y0], args=(r,), t_eval=t_eval, method='RK45')
    plt.plot(sol.t, sol.y[0], label=fr'$y_0={y0}$')

plt.xlabel('$t$')
plt.ylabel('$y(t)$')
plt.title(r"Numerical solution of $\frac{dy}{dt} = r (y-1)(2-y)(y-3)$ for $r=2$")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```
